```python
from datasets import load_dataset

configs = ["en_AU", "en_IN", "en_UK"]

datasets = {}

for config in configs:
    datasets[config] = load_dataset(
        "unswnlporg/BESSTIE",
        config
    )

    print(config)
    print(datasets[config])
```

```python
for config, data in datasets.items():

    for split in data:
        df = data[split].to_pandas()

        filename = f"BESSTIE_{config}_{split}.csv"

        df.to_csv(filename, index=False)

        print(f"Saved: {filename}")
```
