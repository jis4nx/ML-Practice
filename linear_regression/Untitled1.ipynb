```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = make_regression(n_samples=1000, n_features=10, n_targets=1, noise=20,random_state=10)
X_train,X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=10)
```

```python
class GDRegressor:
    def __init__(self, epochs=100, learning_rate=0.01):
        self._coef = []
        self._intercept = 0
        self.lr = learning_rate
        self.epochs = epochs

    
    def fit(self,X,y):
        self._coef = np.full(X.shape[1], 1)
        for i in range(self.epochs):
            
            y_pred = self._intercept + np.dot(X, self._coef)
            
            db = -2 * np.mean(y - y_pred)
            dm = -2 * np.dot((y - y_pred), X)/X.shape[0]
            self._intercept = self._intercept - (self.lr * db)
            self._coef = self._coef - (self.lr * dm)

        return self._intercept, self._coef

    def predict(self,X_test):
        return self._intercept + np.dot(X_test,self._coef)
```

```python
import time
```

```python
start = time.perf_counter()
gr = GDRegressor()
gr.fit(X_train, y_train)
end = time.perf_counter()
gr.predict(X_test)
print(f"Elapsed time: {end - start:.6f} seconds")
```
