```python
import numpy as np
from sklearn.metrics import r2_score
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
X,y = make_regression(n_samples=1000000, n_features=10, n_targets=1, noise=20, random_state=10)
```

```python
class SGDRegressor:
    def __init__(self, epochs=100, learning_rate=0.01):
        self._coef = []
        self._intercept = 0
        self.lr = learning_rate
        self.epochs = epochs

    
    def fit(self,X,y):
        self._coef = np.full(X.shape[1], 1)
        for i in range(self.epochs):
            perm = np.random.permutation(X.shape[0])
            X_perm, y_perm = X[perm], y[perm]

            for idx in range(X.shape[0]):
                y_pred = self._intercept + np.dot(X_perm[idx], self._coef)

                error = y_perm[idx] - y_pred
                db = -2 * error
                dm = -2 * error * X_perm[idx]
                self._intercept = self._intercept - (self.lr * db)
                self._coef = self._coef - (self.lr * dm)

        return self._intercept, self._coef

    def predict(self,X_test):
        return self._intercept + np.dot(X_test,self._coef)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=10)
```

```python
import time
```

```python
start = time.perf_counter()
sgd = SGDRegressor(epochs=1)
sgd.fit(X_train, y_train)
y_pred = sgd.predict(X_test)
end = time.perf_counter()
print(f"Elapsed time: {end - start:.6f} seconds")
```
